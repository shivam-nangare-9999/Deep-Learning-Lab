```python
import numpy as np
```

```python
#Activation function
def relu(x):
  return np.maximum(0,x)
def sigmoid(x):
  return 1/ (1*np.exp(-x))
```

```python
#input Data
x = np.array([2,3])
```

```python
# Initialize weight and biases
w1 = np.array([[0.5, 1.2],
               [0.8,0.6]])
b1 = np.array([1,1])
w2 = np.array([[0.3],
               [0.9]])
b2 = np.array([1])
```

```python
# Hidden layer computation
z1 = np.dot(x,w1)+b1
a1 = relu(z1)
```

```python
# Output layer computation
z2 = np.dot(a1, w2) + b2
Output = sigmoid(z2)
```

```python
# print result
print("input",x)
print("hidden layer weighted sum(z1):",z1)
print("hidden layer activation(a1)",a1)
print("output layer weighted sum (z2),z2")
print("final output",Output)
```
