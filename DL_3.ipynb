```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Activation Function
```

```python
def sigmoid(x):
    return 1 / (1+np.exp(-x))
```

```python
def tanh(x):
    return np.tanh(x)
```

```python
def relu(x):
    return np.maximum(0,x)
```

```python
def leaky_relu(x , alpha=0.01):
    return np.where(x > 0, x, alpha*x)
```

```python
def softmax(x):
    exp_x = np.exp(x - np.max(x))
    return exp_x / np.sum(exp_x)
```

```python
x = np.linspace(-10 , 10, 400)
```

```python
x[250]
```

```python
plt.figure()
plt.plot(x, sigmoid(x))
plt.title("Sigmoid Activation Function")
plt.xlabel("x")
plt.ylabel("Sigmoid(x)")
plt.show()
```

```python
plt.figure()
plt.plot(x, tanh(x))
plt.title("Tanh Activation Function")
plt.xlabel("x")
plt.ylabel("Tanh(x)")
plt.show()
```

```python
plt.figure()
plt.plot(x, relu(x))
plt.title("ReLU Activation Function")
plt.xlabel("x")
plt.ylabel("ReLU(x)")
plt.show()
```

```python
plt.figure()
plt.plot(x, leaky_relu(x))
plt.title("Leaky ReLU Activation Function")
plt.xlabel("x")
plt.ylabel("Leaky ReLU(x)")
plt.show()
```

```python
x_soft = np.array([1.0, 2.0, 3.0, 4.0])
softmax_output = softmax(x_soft)
```

```python
plt.figure()
plt.bar(range(len(x_soft)), softmax_output)
plt.title("Softmax Activation Function")
plt.xlabel("Class Index")
plt.ylabel("Probability")
plt.show()
```

```python
def mean_squared_error(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)
```

```python
y_true = np.array([1, 0, 1, 0])
y_pred = np.array([0.9, 0.1, 0.8, 0.2])

mse_loss = mean_squared_error(y_true, y_pred)
print("Mean Squared Error:", mse_loss)
```

```python
def categorical_crossentropy(y_true, y_pred):
    y_pred = np.clip(y_pred, 1e-9, 1 - 1e-9)
    return -np.sum(y_true * np.log(y_pred))

y_true_ce = np.array([0, 0, 1])
y_pred_ce = softmax(np.array([1.2, 0.9, 2.1]))
ce_loss = categorical_crossentropy(y_true_ce, y_pred_ce)
print("Categorical Cross-Entropy Loss:", ce_loss)
```
